```python
def geneset_exp(adata, gene_set):#to get gene exps in a gene set for all cells from the data (like a vector list for one cell)
    indices = [adata.var_names.get_loc(gene) for gene in list(gene_set.keys()) if gene in adata.var_names] #these are indices from variables(genes) in a gene set.
    #print(indices)
    geneset_exp_matrix = adata.X[:, indices] # to get exps form adata.X(exp array) with their indices
    names = [adata.var_names[idx] for idx in indices] #to get identity of genes 
    if scipy.sparse.issparse(geneset_exp_matrix): #if we have sparse matrix, covering array it.
        geneset_exp_matrix = geneset_exp_matrix.toarray()
    return names,geneset_exp_matrix # this return is called in calc_geneset_score function
```

```python
def edge_weights(gene_set, genes):
    epsilon = 1e-16  # ilişkisi olmayan genler için küçük ağırlık
    results = []

    for gene in genes:  # sadece ekspresyonunu aldığımız genler
        sub_gene_items = gene_set.get(gene, {}).items()

        if not sub_gene_items:  # ilişkisi yoksa epsilon
            N = epsilon
        else:
            sub_genes, weights = zip(*sub_gene_items)
            edge_counts = np.array([len(gene_set.get(g, {})) for g in sub_genes])
            weights = np.array(weights)
            N = np.dot(edge_counts, weights)

        results.append(N)

    return np.array(results)
```

```python
def calc_geneset_score(gene_set, adata): #to take a gene set and calculate the score for all cells
    size=len(gene_set.keys())#just to control the length of a gene set sometimes, I didnt use it.it can be used to normalize scores.
    genes,exps=geneset_exp(adata,gene_set) #getting genes names and exps
    geneset_score=np.dot(exps,edge_weights(gene_set,genes)) #getting edge_weight score and dot multplication to reach gene set score,it is an array which includes cell scores.
    return geneset_score,size
```

```python
def getting_genesets(genesets_relations, adata): #this fucntion calls calc_geneset_score so it calls exps and edge_weight score then append a list 
    genesets_scores = [] 

    genesets_scores = np.array([
    calc_geneset_score(genesets_relations[gene_set], adata)[0] 
    for gene_set in list(genesets_relations.keys())[:]])

    return genesets_scores.T
```

```python
import scanpy as sc
from scipy.stats import ttest_ind, ranksums

group1 = adata_scores[adata_scores.obs["is_tan"]]
group2 = adata_scores[adata_scores.obs["is_nan"]]

t_stat, p_vals = ttest_ind(group1.X, group2.X, axis=0)

w_stat, p_vals_wilcoxon = ranksums(group1.X.mean(axis=1), group2.X.mean(axis=1))
```

```python
# 1. NaN'leri 1 yapıyoruz (yani anlamlı sonuç vermemiş ama analizi bozmaması için)
p_vals = np.nan_to_num(p_vals, nan=1.0)

# 2. 0 olanları çok küçük bir pozitif değere çekiyoruz (istatistiksel olarak 0'a çok yakın, ama hesaplamaya uygun)
p_vals[p_vals == 0] = 1e-300
```

```python
from statsmodels.stats.multitest import multipletests

p_adjusted =multipletests(p_vals, method='fdr_bh')[1]
```

```python
import pandas as pd
import numpy as np
from tabulate import tabulate

# Mean of gene sets
group1_mean = np.array(group1.X.mean(axis=0)).flatten()
group2_mean = np.array(group2.X.mean(axis=0)).flatten()

# Create DataFrame
gene_comparison = pd.DataFrame({
    'TAN Mean Act.': group1_mean,
    'NAN Mean Act': group2_mean,
    'Score Difference': group1_mean - group2_mean,
    't-test': t_stat,
    'p_value': p_vals,
    'p_adjusted': p_adjusted,
    '#of gene in GS': group1.var['gene_size']
})

# Sort by expression difference
gene_comparison = gene_comparison.sort_values(by='Score Difference', ascending=False)

# Display function for formatted output
def print_gene_sets(title, df, condition, ascending=True):
    filtered_df = df[condition].sort_values(by='Score Difference', ascending=ascending)
    if not filtered_df.empty:
        print(f"\n{title}:")
        print(tabulate(filtered_df.head(10), headers='keys', tablefmt='pretty', showindex=True))  # Show top 10 entries

# Print highly expressed gene sets
print_gene_sets("Gene Sets Highly Expressed in TAN Groups", gene_comparison, gene_comparison['Score Difference'] > 0, ascending=False)
print_gene_sets("Gene Sets Highly Expressed in NAN Groups", gene_comparison, gene_comparison['Score Difference'] < 0, ascending=True)

#yanına p-adjusted t test koy , difference a göre rank et. Gsea analizi yap karşılaştır.normalized enrichment score baz alınacak
#gen setleri filtrelenmesin dikkat et gsea da
#performans testi paket için.( hücre sayısı ve gen seti sayısından etkileniyor mu) grafik sunulacak.
```
